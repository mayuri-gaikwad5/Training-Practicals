# 7. HTTP MESSAGE BODY

## 1. Simple Meaning

The HTTP message body contains the actual data or content
that is transferred between the client and the server.

The body can be present in both:

    HTTP Request

and

    HTTP Response


## 2. Basic Structure

                HTTP MESSAGE
                     │
          ┌──────────┴──────────┐
          │                     │
       HEADERS                BODY
          │                     │
          ▼                     ▼
    Information ABOUT       Actual DATA
       the message          or CONTENT


Example:

    Content-Type: application/json

    {
        "name": "Mayuri",
        "age": 21
    }

Here:

    Content-Type: application/json
          ↓
       Header

    {
        "name": "Mayuri",
        "age": 21
    }
          ↓
       Message Body


## 3. Message Body in a Request

The client can send data to the server using
the request body.

Example:

    POST /users HTTP/1.1
    Host: example.com
    Content-Type: application/json

    {
        "name": "Mayuri",
        "age": 21
    }


Visual:

    CLIENT
       │
       │
       │  HTTP REQUEST
       │
       │  POST /users
       │
       │  Headers
       │
       │  Body
       │  {
       │      "name": "Mayuri"
       │  }
       ▼
    SERVER


The body contains the data being sent to the server.


## 4. Message Body in a Response

The server can also send data back to the client
using the response body.

Example:

    HTTP/1.1 200 OK
    Content-Type: application/json

    {
        "id": 1,
        "name": "Mayuri"
    }


Visual:

    SERVER
       │
       │  HTTP RESPONSE
       │
       │  Status Line
       │
       │  Headers
       │
       │  Body
       │  {
       │      "id": 1,
       │      "name": "Mayuri"
       │  }
       ▼
    CLIENT


## 5. Complete HTTP Message

    ┌────────────────────────────────────┐
    │ POST /users HTTP/1.1               │
    │ Host: example.com                  │
    │ Content-Type: application/json     │
    ├────────────────────────────────────┤
    │                                    │
    │ Blank Line                         │
    │                                    │
    ├────────────────────────────────────┤
    │ {                                  │
    │     "name": "Mayuri",              │
    │     "age": 21                      │
    │ }                                  │
    └────────────────────────────────────┘
                    │
                    ▼
               Message Body


## 6. Why Are Headers Needed?

The receiver needs to understand what the body contains.

For example:

    Content-Type: application/json

tells the receiver:

    "The body contains JSON data."


Another example:

    Content-Type: text/html

means:

    "The body contains HTML."


Therefore:

    HEADER
       ↓
    Describes the content


    BODY
       ↓
    Contains the content


## 7. Types of Data in a Message Body

The body can contain different types of data.

Examples:

    JSON
    HTML
    XML
    Plain Text
    Form Data
    Images
    Audio
    Video
    Files / Binary Data


Visual:

                  MESSAGE BODY
                       │
        ┌──────────────┼──────────────┐
        │              │              │
        ▼              ▼              ▼
       Text          Structured      Binary
        │               Data            Data
        │               │               │
        ▼               ▼               ▼
      HTML            JSON            Image
      Text            XML             File
                                     Audio


## 8. Is Message Body Always Present?

No.

The message body is optional in many HTTP messages.

For example:

    GET /posts/1 HTTP/1.1
    Host: example.com


This request may contain no body.

But:

    POST /users HTTP/1.1
    Content-Type: application/json

    {
        "name": "Mayuri"
    }


contains a body.


## 9. GET vs POST Example

### GET

    GET /users/101 HTTP/1.1
    Host: example.com


Usually:

    GET
      ↓
    Request resource
      ↓
    No request body required


### POST

    POST /users HTTP/1.1
    Host: example.com
    Content-Type: application/json

    {
        "name": "Mayuri"
    }


Here:

    POST
      ↓
    Send data
      ↓
    Request body contains the data


IMPORTANT:

GET requests can technically have a body in some
HTTP implementations, but it is not normally used
for ordinary GET requests.


## 10. Blank Line and Body

The blank line separates the headers from the body.

    Headers
       │
       ▼
    Content-Type: application/json
       │
       ▼
    ─────────────────
       BLANK LINE
    ─────────────────
       │
       ▼
    Message Body


So the receiver knows:

    "Headers are finished.
     The content starts here."


## 11. Request Body vs Response Body

    REQUEST BODY

    Client
       │
       │
       │  Data
       ▼
    Server


    RESPONSE BODY

    Server
       │
       │
       │  Data
       ▼
    Client


Simple memory:

    REQUEST BODY
        ↓
    Client sends data


    RESPONSE BODY
        ↓
    Server sends data


## 12. Message Body and Content-Length

A response can contain:

    Content-Length: 55

This tells the receiver the length of the
message body in bytes.

Example:

    Content-Length: 55

    <html>
    <body>
    <h1>First Program</h1>
    </body>
    </html>


Here:

    Content-Length
          ↓
    Information about the body


    HTML
          ↓
    Actual body


## 13. Message Body and Content-Type

Example:

    Content-Type: text/html

    <html>
        <body>
            Hello
        </body>
    </html>


Here:

    Content-Type
          ↓
    Describes the type of body


    HTML
          ↓
    Actual content


Another example:

    Content-Type: application/json

    {
        "id": 1
    }


## 14. Complete Flow

                  CLIENT
                     │
                     │
                     │ HTTP REQUEST
                     │
                     │ Headers
                     │
                     │ Body
                     ▼
                  SERVER
                     │
                     │
                     │ HTTP RESPONSE
                     │
                     │ Headers
                     │
                     │ Body
                     ▼
                  CLIENT


## 15. Quick Memory Trick

    HEADERS
       ↓
    "Information about the data"


    BODY
       ↓
    "Actual data"


    BLANK LINE
       ↓
    "Separates headers from body"


## 16. One-Line Summary

The HTTP message body contains the actual data or content
being transferred between the client and server, while
headers provide information about that content.
