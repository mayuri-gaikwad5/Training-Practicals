# 3. HTTP MESSAGE STRUCTURE




## 1. Simple Meaning

An HTTP message is the complete message exchanged between
a client and a server.

There are two types:

    1. HTTP Request
    2. HTTP Response


## 2. Basic Communication

                  HTTP COMMUNICATION

        CLIENT                         SERVER
           │                              │
           │────── HTTP REQUEST ─────────>│
           │                              │
           │<───── HTTP RESPONSE ─────────│
           │                              │


## 3. HTTP Request Structure

An HTTP request is sent by the client to the server.

    HTTP REQUEST
         │
         ├── Request Line
         │
         ├── Headers
         │
         ├── Blank Line
         │
         └── Optional Message Body


Visual:

    ┌─────────────────────────────────┐
    │         Request Line             │
    ├─────────────────────────────────┤
    │         Request Headers         │
    │         Header: Value            │
    │         Header: Value            │
    ├─────────────────────────────────┤
    │         Blank Line              │
    ├─────────────────────────────────┤
    │         Message Body            │
    │         (Optional)              │
    └─────────────────────────────────┘


## 4. HTTP Response Structure

A response is sent by the server back to the client.

    HTTP RESPONSE
         │
         ├── Status Line
         │
         ├── Headers
         │
         ├── Blank Line
         │
         └── Optional Message Body


Visual:

    ┌─────────────────────────────────┐
    │          Status Line             │
    ├─────────────────────────────────┤
    │          Response Headers       │
    │          Header: Value            │
    │          Header: Value            │
    ├─────────────────────────────────┤
    │          Blank Line              │
    ├─────────────────────────────────┤
    │          Message Body            │
    │          (Optional)              │
    └─────────────────────────────────┘


## 5. Complete HTTP Message Flow

                     HTTP MESSAGE
                           │
              ┌────────────┴────────────┐
              │                         │
           REQUEST                   RESPONSE
              │                         │
              ▼                         ▼
        Request Line               Status Line
              │                         │
              ▼                         ▼
          Headers                    Headers
              │                         │
              ▼                         ▼
         Blank Line                 Blank Line
              │                         │
              ▼                         ▼
        Optional Body             Optional Body


## 6. Request Example

    GET /posts/1 HTTP/1.1
    Host: example.com
    Accept: application/json

    [Optional Body]


Here:

    GET /posts/1 HTTP/1.1
              ↓
         Request Line

    Host: example.com
              ↓
         Request Header

    Accept: application/json
              ↓
         Request Header

    Blank line
              ↓
    Separates headers from body


## 7. Response Example

    HTTP/1.1 200 OK
    Content-Type: application/json
    Content-Length: 45

    {
        "id": 1,
        "title": "Learning HTTP"
    }


Here:

    HTTP/1.1 200 OK
              ↓
         Status Line

    Content-Type: application/json
              ↓
         Response Header

    Content-Length: 45
              ↓
         Response Header

    JSON data
              ↓
         Message Body


## 8. What is the Blank Line?

The blank line is very important.

It separates:

    Headers
       ↓
    Blank Line
       ↓
    Message Body


Visual:

    Headers
       │
       │
       ▼
    ───────────
    BLANK LINE
    ───────────
       │
       ▼
    Message Body


It tells the receiver:

    "The headers are finished.
     If a body exists, it starts after this."


## 9. What is a Message Body?

The message body contains the actual data/content
being transferred.

Examples:

    JSON
    HTML
    XML
    Plain Text
    Form Data
    File/Binary Data


Example:

    Content-Type: application/json

    {
        "name": "Mayuri",
        "age": 21
    }


Here:

    Headers
       ↓
    describe the content

    Body
       ↓
    contains the actual content


## 10. Is the Body Always Present?

No.

A message body is optional.

For example, a typical GET request may be:

    GET /posts/1 HTTP/1.1
    Host: example.com


There is no body here.

But a POST request commonly contains a body:

    POST /users HTTP/1.1
    Content-Type: application/json

    {
        "name": "Mayuri"
    }


## 11. Request vs Response

    REQUEST

    Client ───────────────> Server

    Purpose:
    Tell the server what the client wants
    or send data to the server.


    RESPONSE

    Server ───────────────> Client

    Purpose:
    Tell the client what happened
    and optionally return data.


## 12. Important Components

    REQUEST
       │
       ├── Request Line
       ├── Request Headers
       ├── Blank Line
       └── Optional Body


    RESPONSE
       │
       ├── Status Line
       ├── Response Headers
       ├── Blank Line
       └── Optional Body


## 13. Quick Memory Trick

    REQUEST

    "What do I want?"

         ↓

    Request Line
    + Headers
    + Body


    RESPONSE

    "What happened?"

         ↓

    Status Line
    + Headers
    + Body


## 14. One-Line Summary

An HTTP message consists of a start line, headers,
a blank line, and an optional message body.
A request uses a request line, while a response uses
a status line.
