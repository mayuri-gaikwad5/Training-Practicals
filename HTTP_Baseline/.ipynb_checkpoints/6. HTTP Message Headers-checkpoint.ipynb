# 6. HTTP MESSAGE HEADERS

## 1. Simple Meaning

HTTP headers provide additional information about an
HTTP request or response.

They are written in the form:

    Header-Name: Value


Example:

    Host: example.com

    Content-Type: application/json

    Accept: text/html


Think of headers as:

    "Information ABOUT the HTTP message or its content."


## 2. Where Are Headers Present?

Headers can be present in both:

                    HTTP MESSAGE
                         │
             ┌───────────┴───────────┐
             │                       │
          REQUEST                 RESPONSE
             │                       │
             ▼                       ▼
      Request Headers          Response Headers


## 3. Basic HTTP Message Structure

    ┌─────────────────────────────────┐
    │ Start Line                      │
    ├─────────────────────────────────┤
    │ Header: Value                   │
    │ Header: Value                   │
    │ Header: Value                   │
    ├─────────────────────────────────┤
    │ Blank Line                      │
    ├─────────────────────────────────┤
    │ Optional Message Body           │
    └─────────────────────────────────┘


## 4. Four Traditional Types of Headers

Older HTTP/1.1 terminology classifies headers into:

                    HTTP HEADERS
                         │
          ┌──────────────┼──────────────┐
          │              │              │
          ▼              ▼              ▼
       GENERAL        REQUEST        RESPONSE
       HEADERS        HEADERS        HEADERS
                         │
                         │
                         ▼
                   ENTITY HEADERS
                  (Older terminology)


The exact classification above comes from older
HTTP/1.1 terminology.


## 5. General Headers

General headers can be used with both requests
and responses.

They provide information about the HTTP message
or communication rather than a specific request
or response alone.

Examples include:

    Cache-Control
    Connection


Example:

    Cache-Control: max-age=3600


The same general concept can apply to both
requests and responses.


## 6. Request Headers

Request headers are sent from the client to the server.

They provide information about:

    • The client
    • The requested resource
    • Client preferences
    • Authentication
    • Other request-related information


Example:

    GET /posts/1 HTTP/1.1
    Host: example.com
    Accept: application/json
    User-Agent: Mozilla/5.0


Here:

    Host
       ↓
    Identifies the requested host


    Accept
       ↓
    Tells the server which response format
    the client prefers


    User-Agent
       ↓
    Identifies the client software


Visual:

    CLIENT
       │
       │ Request Headers
       │
       ├── Host
       ├── Accept
       └── User-Agent
       │
       ▼
    SERVER


## 7. Response Headers

Response headers are sent from the server
to the client.

They provide information about:

    • The response
    • The server
    • The returned content
    • Caching
    • Other response-related information


Example:

    HTTP/1.1 200 OK
    Content-Type: application/json
    Content-Length: 45
    Server: Apache


Here:

    Content-Type
       ↓
    Describes the type of returned content


    Content-Length
       ↓
    Gives the size of the message body


    Server
       ↓
    Identifies the server software


Visual:

    SERVER
       │
       │ Response Headers
       │
       ├── Content-Type
       ├── Content-Length
       └── Server
       │
       ▼
    CLIENT


## 8. Entity Headers

"Entity headers" is an older HTTP/1.1 term.

They provide metadata about the entity body
or content.

Examples:

    Content-Type
    Content-Length
    Last-Modified


Example:

    Content-Type: text/html

This tells the receiver:

    "The content is HTML."


Example:

    Content-Length: 55

This tells the receiver:

    "The content has a length of 55 bytes."


## 9. What is Metadata?

Metadata means:

    "Information about the data."


Example:

    Actual data:

    <h1>Hello</h1>


    Metadata:

    Content-Type: text/html


The HTML is the actual content.

The Content-Type describes the content.


Visual:

              DATA
               │
       ┌───────┴────────┐
       │                │
       ▼                ▼
    Metadata          Content
       │                │
       ▼                ▼
 Content-Type       <h1>Hello</h1>


## 10. Complete Example

    GET /first.htm HTTP/1.1
    Host: www.example.com
    User-Agent: Mozilla/5.0
    Accept-Language: en-us
    Accept-Encoding: gzip


Here:

    Host
       ↓
    Request Header

    User-Agent
       ↓
    Request Header

    Accept-Language
       ↓
    Request Header

    Accept-Encoding
       ↓
    Request Header


The server may respond:

    HTTP/1.1 200 OK
    Date: Wed, 4 Dec 2019 05:15:40 GMT
    Server: Apache/2.4.41
    Content-Length: 55
    Content-Type: text/html


Here:

    Date
       ↓
    Response Header

    Server
       ↓
    Response Header

    Content-Length
       ↓
    Response / content metadata

    Content-Type
       ↓
    Response / content metadata


## 11. Header Structure

Every header generally follows:

    Header-Name: Value


Example:

    Content-Type: text/html
    │            │
    │            └── Value
    └─────────────── Header Name


Another example:

    Accept-Language: en-us


    Accept-Language
          ↓
      Header Name

    en-us
          ↓
        Value


## 12. Headers and Message Body

Headers describe or provide information about
the message and its content.

The body contains the actual content.

    HTTP MESSAGE
         │
         ├── Headers
         │     ↓
         │   "Information about
         │    the message/content"
         │
         └── Body
               ↓
            "Actual content"


Example:

    Content-Type: application/json

    {
        "name": "Mayuri"
    }


Here:

    Content-Type
        ↓
    Describes the body


    JSON object
        ↓
    Actual body/content


## 13. Quick Comparison

    GENERAL HEADER
         ↓
    General information about
    HTTP communication/message


    REQUEST HEADER
         ↓
    Client → Server


    RESPONSE HEADER
         ↓
    Server → Client


    ENTITY HEADER
         ↓
    Older term for metadata
    about the entity/content


## 14. Important Modern Note

"Entity header" is an older HTTP/1.1 term.

Modern HTTP specifications generally use terms such as:

    Message Content
    Representation Metadata

For syllabus purposes, understand:

    Entity Header
         ↓
    Metadata describing the content/entity


## 15. Quick Memory Trick

    REQUEST HEADERS

    Client → Server

         ↓

    "Here is information
     about my request."


    RESPONSE HEADERS

    Server → Client

         ↓

    "Here is information
     about my response."


    ENTITY HEADERS

         ↓

    "Here is information
     about the content."


## 16. One-Line Summary

HTTP headers are key-value pairs that provide additional
information about an HTTP request, response, or its content.
Older HTTP/1.1 terminology classifies them as general,
request, response, and entity headers.
