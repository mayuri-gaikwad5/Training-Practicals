# HTTP PARAMETERS

## 1. Simple Meaning

HTTP parameters are additional pieces of information sent along with an HTTP request to help the server understand what data is required or what action should be performed.

A parameter generally has:

    Key = Value

Example:

    userId = 1

The parameter tells the server:

    "Give me the data related to user 1."

## 2. Where Can Information Be Sent?

HTTP request information can commonly appear in:

                HTTP REQUEST
                     │
          ┌──────────┼───────────┐
          │          │           │
          ▼          ▼           ▼
       URL/Path    Headers      Body
          │
          ▼
    Query Parameters
    Path Parameters

## 3. Query Parameters

Query parameters are parameters added to the URL after a '?'.

Example:

    https://example.com/posts?userId=1

Structure:

    https://example.com/posts?userId=1
    │        │          │      │    │
    │        │          │      │    └── Value
    │        │          │      └─────── Key
    │        │          └────────────── Query parameter
    │        └───────────────────────── Path
    └────────────────────────────────── Domain


### Query Parameter Syntax

    ?key=value


Example:

    ?userId=1


Multiple query parameters:

    ?userId=1&limit=10

Here:

    userId = 1
    limit  = 10

The '&' separates multiple query parameters.

## 4. Path Parameters

A path parameter is included directly in the URL path and is generally used to identify a specific resource.

Example:

    https://example.com/users/101

Here:

    /users/101

The "101" identifies a particular user.

Visual:

    /users/101
       │   │
       │   └── Specific user
       └────── Resource collection


Another example:

    /posts/25

This can mean:

    "The post whose ID is 25."

## 5. Query Parameter vs Path Parameter

    Query Parameter:

    /posts?userId=1

    Meaning:

    "Give me posts according to this condition/filter."


    Path Parameter:

    /posts/1

    Meaning:

    "Give me the specific post identified by 1."


Visual:

                  URL
                   │
          ┌────────┴────────┐
          │                 │
        Path              Query
          │                 │
          ▼                 ▼
      /posts/1       /posts?userId=1
          │                 │
          ▼                 ▼
   Identify resource     Filter/modify
   using path            the request

## 6. Request Body

Data can also be sent inside the HTTP message body.

Example:

    POST /users HTTP/1.1

    Content-Type: application/json

    {
        "name": "Mayuri",
        "age": 21
    }

Here the data is not present in the URL.

The body contains the data that is being sent to the server.


Visual:

    Client
      │
      │  POST /users
      │
      │  Body:
      │  {
      │      "name": "Mayuri",
      │      "age": 21
      │  }
      ▼
    Server


## 7. HTTP Headers

Headers carry additional information about the HTTP request or response.

Example:

    Authorization: Bearer abc123

    Accept: application/json

Headers are different from query/path parameters.

Visual:

    HTTP REQUEST
         │
         ├── Request Line
         │
         ├── Headers
         │      ├── Host
         │      ├── Accept
         │      └── Authorization
         │
         └── Body
                │
                └── Actual data


## 8. Complete Example

Request:

    POST /users?source=mobile HTTP/1.1
    Host: example.com
    Content-Type: application/json

    {
        "name": "Mayuri",
        "age": 21
    }


Breaking it down:

    /users
        ↓
    Path

    source=mobile
        ↓
    Query Parameter

    Host
        ↓
    Header

    JSON data
        ↓
    Message Body

## 9. Key Difference

    Path Parameter
        ↓
    Usually identifies WHICH resource

    Query Parameter
        ↓
    Usually provides filtering, searching,
    sorting, pagination, or other options

    Header
        ↓
    Provides metadata/instructions about
    the HTTP request

    Body
        ↓
    Carries data/content being sent


## 10. Important Note

"HTTP parameter" is a broad/general term.

Query parameters, path parameters, body data, and headers are NOT all technically the same type of parameter.

They are different ways of carrying information in an HTTP request.


## 11. Quick Memory Trick

    PATH
      ↓
    "Which resource?"

    QUERY
      ↓
    "What conditions/options?"

    HEADER
      ↓
    "Information about the request"

    BODY
      ↓
    "What data am I sending?"


## 12. One-Line Summary

HTTP parameters and related request information help the client provide the server with details about the requested resource, preferences, options, and data.
