# 5. HTTP RESPONSE / STATUS LINE

## 1. Simple Meaning

The Status Line is the first line of an HTTP response.

It tells the client:

    1. Which HTTP version is being used
    2. What happened to the request
    3. A short description of the result


## 2. Basic Syntax

    HTTP-VERSION    STATUS-CODE    REASON-PHRASE

Example:

    HTTP/1.1 200 OK


Visual:

    HTTP/1.1        200          OK
       │             │            │
       │             │            └── Reason Phrase
       │             └─────────────── Status Code
       └───────────────────────────── HTTP Version


## 3. Three Parts of Status Line

                    STATUS LINE
                         │
          ┌──────────────┼──────────────┐
          │              │              │
          ▼              ▼              ▼
    HTTP-VERSION    STATUS-CODE    REASON-PHRASE
          │              │              │
          ▼              ▼              ▼
       HTTP/1.1          200            OK


## 4. HTTP Version

The first part specifies the HTTP version used
for the response.

Example:

    HTTP/1.1 200 OK
    └───────┘
    HTTP Version


Examples:

    HTTP/1.0
    HTTP/1.1


## 5. Status Code

The status code is a three-digit number that tells
the client the result of the request.

Example:

    HTTP/1.1 200 OK
             └───┘
           Status Code


Common status codes:

    200 → OK
    201 → Created
    301 → Moved Permanently
    302 → Found
    400 → Bad Request
    401 → Unauthorized
    403 → Forbidden
    404 → Not Found
    500 → Internal Server Error


## 6. Status Code Categories

Status codes are grouped according to their
first digit.

                    STATUS CODES
                         │
          ┌──────────────┼──────────────┐
          │              │              │
          ▼              ▼              ▼
        1xx            2xx            3xx
       Informational   Success       Redirection
          
          │
          ├──────────────────────────────┐
          │                              │
          ▼                              ▼
        4xx                            5xx
       Client Error                  Server Error


## 7. Important Categories

    1xx
      ↓
    Informational response


    2xx
      ↓
    Request was successfully processed


    3xx
      ↓
    Redirection / further action required


    4xx
      ↓
    Problem with the request from the client side


    5xx
      ↓
    Server encountered a problem processing
    a valid request


## 8. Reason Phrase

The reason phrase is a short human-readable
description of the status code.

Example:

    HTTP/1.1 200 OK
                  └──┘
             Reason Phrase


Examples:

    200 OK
    404 Not Found
    500 Internal Server Error


The status code carries the actual protocol meaning.

The reason phrase provides a readable description.


## 9. Complete Example

    HTTP/1.1 200 OK


Breaking it down:

    HTTP/1.1
       ↓
    HTTP Version


    200
       ↓
    Status Code
       ↓
    Request was successful


    OK
       ↓
    Reason Phrase
       ↓
    Human-readable description


## 10. Another Example

Suppose the client requests:

    GET /posts/999 HTTP/1.1
    Host: example.com


If the requested post does not exist,
the server may respond:

    HTTP/1.1 404 Not Found


Flow:

    Client
       │
       │ GET /posts/999
       ▼
    Server
       │
       │ Resource not found
       ▼
    HTTP/1.1 404 Not Found
       │
       ├── HTTP/1.1 → Version
       ├── 404      → Status Code
       └── Not Found → Reason Phrase


## 11. Request Line vs Status Line

These two are very important to distinguish.

    REQUEST LINE

    Client → Server

    GET /posts/1 HTTP/1.1

    Means:

    "What does the client want?"


    STATUS LINE

    Server → Client

    HTTP/1.1 200 OK

    Means:

    "What happened to the request?"


Visual:

    CLIENT                         SERVER
       │                              │
       │  GET /posts/1 HTTP/1.1       │
       │─────────────────────────────>│
       │                              │
       │  HTTP/1.1 200 OK             │
       │<─────────────────────────────│
       │                              │


## 12. Status Line in HTTP Response

    ┌─────────────────────────────────────┐
    │ HTTP/1.1 200 OK                     │
    │                                     │
    │ Content-Type: application/json      │
    │ Content-Length: 45                  │
    │                                     │
    │                                     │
    │ {                                   │
    │   "id": 1                           │
    │ }                                   │
    └─────────────────────────────────────┘
          │
          ▼
    First line = Status Line


## 13. Quick Memory Trick

    REQUEST LINE

        "What do I want?"

             ↓

    GET /posts/1 HTTP/1.1


    STATUS LINE

        "What happened?"

             ↓

    HTTP/1.1 200 OK


## 14. One-Line Summary

The HTTP Status Line is the first line of an HTTP
response and contains the HTTP version, status code,
and reason phrase.
