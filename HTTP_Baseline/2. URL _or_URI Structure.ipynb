# 2. URL / URI STRUCTURE

## 1. Simple Meaning

A URL (Uniform Resource Locator) tells us:

    "Where is the resource and how can I access it?"

A URI (Uniform Resource Identifier) identifies a resource.

Example:

    https://example.com/posts/101

## 2. Basic Structure

    https://example.com/posts/101?userId=5
    │       │           │        │
    │       │           │        └── Query Parameter
    │       │           └─────────── Path
    │       └─────────────────────── Domain / Host
    └─────────────────────────────── Scheme / Protocol

## 3. Scheme / Protocol

    https://

The scheme tells the client which protocol should be used
to communicate with the server.

Common examples:

    http://
    https://


HTTPS is the secure version of HTTP.

## 4. Domain / Host

    example.com

The domain identifies the host/server that the client
wants to communicate with.

Example:

    https://example.com/posts

Here:

    example.com
          ↓
       Domain/Host


IMPORTANT:

The domain is NOT the resource.

It identifies where the request is being sent.

## 5. Path

    /posts/101

The path identifies the resource or resource path
that the client wants to access.

Example:

    /posts

can represent a collection of posts.

While:

    /posts/101

can represent a specific post.


Visual:

    /posts/101
       │   │
       │   └── Specific resource
       └────── Resource collection

## 6. Query String

A query string starts with:

    ?

Example:

    ?userId=5

The query string contains query parameters.

Structure:

    ?key=value

Example:

    ?userId=5

    key   = userId
    value = 5


Multiple query parameters:

    ?userId=5&limit=10

Here:

    userId = 5
    limit  = 10

The '&' separates multiple parameters.


## 7. Complete URL Example

    https://example.com/posts/101?userId=5


    https://
       ↓
    Scheme

    example.com
       ↓
    Domain / Host

    /posts/101
       ↓
    Path

    ?userId=5
       ↓
    Query Parameter


## 9. URL vs URI

URI stands for:

    Uniform Resource Identifier

URL stands for:

    Uniform Resource Locator


Simple way to remember:

    URI
     ↓
    Identifies a resource


    URL
     ↓
    Identifies a resource
    and provides a way/location
    to access it


Therefore:

    Every URL is a URI

but:

    Not every URI is necessarily a URL.

## 10. Resource vs Domain

This is an important distinction.

Example:

    https://example.com/posts/101


    example.com
         ↓
    Domain / Host
         ↓
    Where to send the request


    /posts/101
         ↓
    Path
         ↓
    Which resource/path is requested


Think:

    DOMAIN → "Where?"

    PATH   → "Which resource?"

## 11. Connection With HTTP Request

The complete URL:

    https://example.com/posts/101


can result in an HTTP request such as:

    GET /posts/101 HTTP/1.1
    Host: example.com


Notice:

    Complete URL
          ↓
    https://example.com/posts/101


    HTTP Request
          ↓
    GET /posts/101 HTTP/1.1
    Host: example.com


The domain is generally carried separately in
the Host header in an HTTP/1.1 request.


## 12. Quick Memory Trick

    SCHEME
       ↓
    "How?"

    DOMAIN / HOST
       ↓
    "Where?"

    PATH
       ↓
    "Which resource?"

    QUERY
       ↓
    "What conditions/options?"

## 13. One-Line Summary

A URL identifies the location/access method of a resource,
while a URI is a broader identifier for a resource.
A typical HTTP URL consists of a scheme, domain/host,
path, and optional query parameters.
