# 4. HTTP REQUEST LINE

## 1. Simple Meaning

The Request Line is the first line of an HTTP request.

It tells the server:

    1. What action the client wants
    2. Which resource is being requested
    3. Which HTTP version is being used


## 2. Basic Syntax

    METHOD    REQUEST-TARGET    HTTP-VERSION

Example:

    GET /posts/1 HTTP/1.1


Visual:

    GET        /posts/1        HTTP/1.1
     │             │                │
     │             │                └── HTTP Version
     │             └─────────────────── Request-Target
     └───────────────────────────────── HTTP Method


## 3. Three Parts of Request Line

                    REQUEST LINE
                         │
          ┌──────────────┼──────────────┐
          │              │              │
          ▼              ▼              ▼
       METHOD      REQUEST-TARGET   HTTP-VERSION
          │              │              │
          ▼              ▼              ▼
        GET          /posts/1       HTTP/1.1


## 4. HTTP Method

The method tells the server what action the client
wants to perform.

Common HTTP methods:

    GET
    POST
    PUT
    PATCH
    DELETE


Example:

    GET /posts/1 HTTP/1.1

Here:

    GET
      ↓
    The client wants to retrieve the resource.


Another example:

    POST /users HTTP/1.1

Here:

    POST
      ↓
    The client wants to send data to the server.


## 5. Request-Target

The Request-Target identifies the target of the request.

Example:

    GET /posts/1 HTTP/1.1
        └────────┘
        Request-Target


Here:

    /posts/1
        ↓
    The requested resource/path


The Request-Target is also commonly referred to as
Request-URI in older HTTP/1.1 terminology.


## 6. HTTP Version

The last part specifies the HTTP version being used.

Example:

    GET /posts/1 HTTP/1.1
                       └──────┘
                      HTTP Version


Examples:

    HTTP/1.0
    HTTP/1.1


## 7. Complete Example

    GET /first.htm HTTP/1.1


Breaking it down:

    GET
     ↓
    Method
     ↓
    "Retrieve the resource"


    /first.htm
     ↓
    Request-Target
     ↓
    "The resource being requested"


    HTTP/1.1
     ↓
    HTTP Version
     ↓
    "Use HTTP/1.1"


## 8. Request Line in the HTTP Message

    ┌─────────────────────────────────────┐
    │ GET /posts/1 HTTP/1.1               │
    │                                     │
    │ Host: example.com                   │
    │ Accept: application/json            │
    │                                     │
    │                                     │
    │ Optional Message Body               │
    └─────────────────────────────────────┘
          │
          ▼
    First line = Request Line


## 9. Request Line vs Headers

Do not confuse the Request Line with headers.

    GET /posts/1 HTTP/1.1
             ↓
        Request Line


    Host: example.com
             ↓
        Request Header


    Accept: application/json
             ↓
        Request Header


The Request Line tells the server the main request
information.

Headers provide additional information about the request.


## 10. Request Line and Complete URL

Suppose the URL is:

    https://example.com/posts/1


The HTTP/1.1 request can contain:

    GET /posts/1 HTTP/1.1
    Host: example.com


Notice:

    https://example.com/posts/1
             │          │
             │          └── Path
             └───────────── Domain


In the normal HTTP/1.1 request:

    /posts/1
       ↓
    Request-Target

    example.com
       ↓
    Host Header


## 11. Important Syntax Rule

The three parts are separated by spaces.

    METHOD + SPACE + REQUEST-TARGET + SPACE + HTTP-VERSION


Example:

    GET + SPACE + /posts/1 + SPACE + HTTP/1.1


Correct:

    GET /posts/1 HTTP/1.1


Incorrect:

    GET/posts/1/HTTP/1.1


because the required spaces are missing.


## 12. Quick Memory Trick

    REQUEST LINE

         ↓

    "What do I want?"

         ↓

    METHOD
       +
    "What resource?"
       +
    REQUEST-TARGET
       +
    "Which HTTP version?"
       +
    HTTP-VERSION


## 13. One-Line Summary

The HTTP Request Line is the first line of a request
and contains the HTTP method, request-target, and
HTTP version.
