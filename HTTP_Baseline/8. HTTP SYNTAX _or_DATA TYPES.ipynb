# 8. HTTP SYNTAX / DATA TYPES

HTTP uses certain standard data types and syntax rules in headers and messages.
These help HTTP represent dates, time durations, URIs, character sets,
content encoding, transfer encoding, and software information.

--------------------------------------------------
3. URI
--------------------------------------------------

URI = Uniform Resource Identifier

Simple meaning:
A URI identifies a resource.

Example:

https://example.com/posts/101

Here the URI identifies the resource being accessed.

A URI can contain:

    Scheme
       ↓
    https
       ↓
    Domain
       ↓
    example.com
       ↓
    Path
       ↓
    /posts/101

Another example:

/posts/101

This is a URI reference/path used in an HTTP request.

HTTP request:

GET /posts/101 HTTP/1.1

Here:

/posts/101 → Request-URI / request-target

Remember:

URI = identifies a resource


--------------------------------------------------
7. PRODUCT TOKENS
--------------------------------------------------

Simple meaning:
A product token identifies a software product or version.

Examples:

Mozilla/69.0
Apache/2.4.41
MyApp/1.0

They can appear in headers such as:

User-Agent: MyBrowser/2.0

Server: Apache/2.4.41

Here:

MyBrowser/2.0 → identifies client software

Apache/2.4.41 → identifies server software
